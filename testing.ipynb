```python
arr = [1]

def special(arr) -> bool: 
    for i in range(len(arr)-1): 
        j = i + 1
        if arr[i] % 2 == 0: 
            if arr[j] % 2 == 0: 
                return False
            else: 
                continue
        if arr[i] % 2 != 0: 
            if arr[j] % 2 != 0: 
                return False
            else: 
                continue 

    return True 

print(special(arr))
            
```

```python
digits = [8, 9, 9, 9]

for i in range(len(digits)-1, 0, -1): 
    if digits[i] + 1 > 9: 
        digits[i] = 0 
    else: 
        digits[i] += 1
        break 

if digits[0] + 1 > 9: 
    digits[0] = 0 
    digits = [1] + digits
else: 
    digits[0] += 1
     
print(digits)


    
```
